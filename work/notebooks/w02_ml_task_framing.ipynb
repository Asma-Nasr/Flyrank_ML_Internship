## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This is a ranking problem because refresh-worthiness is a matter of degree, not a category.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

It is a defined-rule proxy, not an observed outcome.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

The defensible metric is precision@k, validated via a proxy or a small human check:

Precision@10 (or @20): of the top-K pages the model surfaces, what fraction would a human editor agree are genuinely worth reviewing?

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

One row = one content page (URL). Each row carries the features you'll score on: ```age_months```, ```position_avg```, ```demand_impressions```, ```clicks_30d```, ```trend_direction```, ```ctr```. The ```opportunity_score``` column would be computed from these and used only to sort the table — the page stays the unit throughout, nothing gets aggregated up to site- or category-level.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A hand-written rule like ```IF trend == "declining" AND position BETWEEN 5 AND 20 THEN flag ```breaks down fast because the real signal is interaction effects across continuous variables, not clean thresholds.
